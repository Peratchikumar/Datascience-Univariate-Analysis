```python
dataset.isna().sum()
```

```python
dataset["salary"].fillna(0,inplace=True)
```

```python
dataset.isnull().sum()
```

```python
dataset=dataset.dropna() #Drop the NAN values
```

```python
dataset
```

```python
dataset["salary"].fillna(dataset["salary"].mean(),inplace=True)
```

```python
dataset
```

```python
def QualitativeQuantitative(dataset): #Function - is a procedure
    Qualitative=[]
    Quantitative=[]
    
    for columnName in dataset.columns: #Print the column name in the dataset - For Loop
        print(columnName)
        if(dataset[columnName].dtype=='object'):
            print("Qualitative")
            Qualitative.append(columnName)
        else:
            print("Quantitative")
            Quantitative.append(columnName)
    return Qualitative, Quantitative    
```

```python
Qualitative,Quantitative=QualitativeQuantitative(dataset)
```

```python
dataset[Qualitative]
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
imp.fit(dataset[Qualitative])
df=imp.transform(dataset[Qualitative])
```

```python
df=pd.DataFrame(df,columns=Qualitative)
```

```python
df
```

```python
cate=dataset[Qualitative] #Categorical Columns
```

```python
cate
```

```python
df=dataset[Quantitative]  #Numerical Columns
```

```python
df
```

```python
two=[cate,df] #How to combine two DataFrames in Pandas
```

```python
preprocessed=pd.concat(two,axis=1)
```
